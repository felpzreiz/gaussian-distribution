### Imports and Dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('SOCR-HeightWeight.csv')
```

### Data Cleaning

```python
df['weight_kilograms'] = df['Weight(Pounds)'] / 2.205

df = (
    df
    .drop(columns=['Height(Inches)', 'Weight(Pounds)'])
    .sort_values(['weight_kilograms'])
)
```

### FDP Parameters

```python
mu = np.mean(df['weight_kilograms'])
sigma = np.std(df['weight_kilograms'])
```

```python
x = np.linspace(mu - 4*sigma, mu + 4*sigma, 1000)

y = (1 / (sigma * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((x - mu) / sigma) ** 2)
```

### Configuring and generating Chart

```python
plt.figure(figsize=(10, 5))


plt.hist(df['weight_kilograms'], bins=30, density=True, alpha=0.6, color='y', edgecolor='black', label="Histograma")
plt.plot(x, y, label="FDP", color='r')
plt.axvline(mu, color='b', linestyle='dashed', label="Média (μ)")


plt.xlabel("Peso (kg)")
plt.ylabel("Densidade de Probabilidade")
plt.title("Distribuição Normal de Pesos (kg)")
plt.legend()

plt.show()
```
