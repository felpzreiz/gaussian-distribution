```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('SOCR-HeightWeight.csv')
```

```python
df['weight_kilograms'] = df['Weight(Pounds)'] / 2.205

df = (
    df
    .drop(columns=['Height(Inches)', 'Weight(Pounds)'])
    .sort_values(['weight_kilograms'])
)
```

```python
plt.hist(df['weight_kilograms'], bins = 'auto' , color = 'y')
plt.show()
```
